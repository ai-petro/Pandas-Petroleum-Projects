```python
import pandas as pd
import numpy as np

#  Read well 15/9-19A log data from local CSV file
Reading = pd.read_csv('c:/Users/TOSHIBA/OneDrive/Desktop/AI/well 15 9-19A.csv')

# Initial inspection: raw data + column summary and data types
print(Reading)
print(Reading.info())

# Check number of missing values per column before cleaning
print(Reading.isnull().sum())

# Clean data: fill missing values with each column's own mean, save result
Reading = Reading.fillna({'GR':Reading['GR'].mean(),'RHOB':Reading['RHOB'].mean(),'NPHI':Reading['NPHI'].mean()})
print(Reading)

# Filter: extract rows with high GR and low density (possible shale indicator) 
print(Reading[(Reading['GR'] > 80) & (Reading['RHOB']>2.30)])

# Create flag column: classify each depth as High_GR or Low_GR based on GR value
Reading['Vsh_flag'] = np.where(Reading['GR']>80 , 'High_GR','Low_GR')
   
# Sort data ascending by depth
print(Reading.sort_values('DEPTH'))

# Compare average GR between High_GR and Low_GR group
print(Reading.groupby('Vsh_flag')['GR'].mean())

# Display final table after all transformations
print(Reading)

```
